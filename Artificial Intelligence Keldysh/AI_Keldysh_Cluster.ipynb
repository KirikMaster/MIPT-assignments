```python
import numpy as np
```

```python
def point_distance(x, y):
    return np.linalg.norm(x - y)

def cluster_distance(X, Y, dist_matrix):
    return min([dist_matrix[p1][p2] for p1 in X for p2 in Y])

def get_dist_matrix(X):
    n = X.shape[0]
    dist_matrix = np.zeros((n, n))
    for i in range(n):
        for j in range(i+1, n):
            dist_matrix[i][j] = point_distance(X[i], X[j])
            dist_matrix[j][i] = dist_matrix[i][j]
    
    return dist_matrix

def hierarchical_method(X):
    n = X.shape[0]
    dist_matrix = get_dist_matrix(X)
    clusters = [{i} for i in range(n)]
    clusters_history = [clusters.copy()]
    while len(clusters) > 1:
        min_dist, min_pair = np.inf, None
        for i in range(len(clusters)):
            for j in range(i+1, len(clusters)):
                c1, c2 = clusters[i], clusters[j]
                dist = cluster_distance(c1, c2, dist_matrix)
                if dist < min_dist:
                    min_dist = dist
                    min_pair = (i, j)
        i, j = min_pair
        newcluster = clusters[i].union(clusters[j])
        clusters[i] = newcluster
        del clusters[j]
        clusters_history.append(clusters.copy())
    return clusters_history
    
def example():
    X = np.array([[1, 2],
              [2, 1],
              [4, 5],
              [5, 4],
              [8, 8]])
    clusters_history = hierarchical_method(X)
    print(clusters_history)
```

```python
example()
```
