### Реализация сети Элмана

```python
class ElmanRNN(torch.nn.Module):
    def __init__(self):
        super(ElmanRNN, self).__init__()
        self.input2hidden = torch.nn.Linear(input_size, hidden_size)
        self.hidden2hidden = torch.nn.Linear(hidden_size, hidden_size, bias=False)
        self.hidden2output = torch.nn.Linear(hidden_size, output_size)
        self.activation = torch.tanh
    def forward(self, x, hidden):
        # Размерность x: (batch_size, seq_length, input_size)
        batch_size, seq_length, input_size = x.size()
        outputs = []
        for i in range(seq_length):
            xi = x[:, i, :]
            hidden = self.activation(self.hidden2hidden(hidden) + self.input2hidden(xi))
            output = self.hidden2output(hidden)
            outputs.append(output) # (batch_size, input_size)
        outputs = torch.stack(outputs, dim=1) # (batch_size, seq_length, output_size)
        return outputs, hidden
    def init_hidden(self, batch_size, hidden_size):
        return torch.zeros((batch_size, hidden_size))
```
