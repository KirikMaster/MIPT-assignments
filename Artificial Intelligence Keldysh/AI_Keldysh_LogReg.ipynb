## Логистическая регрессия

#### 23.10.2024

```python
import numpy as np
```

```python
def sigma(z):
    return 1 / (1 + np.exp(-z))

def predict(X, w, b):
    z = X @ w + b
    y_pred = sigma(z)
    return y_pred

def train(X, y_true, learning_rate=0.1, num_iterations=1000):
    n_samples, n_features = X.shape
    w, b = np.zeros((n_features,)), 0.0
    for i in range(num_iterations):
        z = X @ w + b
        y_pred = sigma(z)
        
        dw = X.T @ (y_pred - y_true) / n_samples
        db = np.sum((y_pred - y_true)) / n_samples
        
        w -= learning_rate * dw
        b -= learning_rate * db
        
        if i % 10 == 0:
            y_class = (y_pred >= 0.5).astype(int)
            accuracy = np.sum(y_class == y_true) / n_samples
            print(f'Итерация {i}. Точность: {accuracy*100:.2f}%')
            
    return w, b

def example():
    
    # 1. Данные
    X = np.array([[100.0, 1.0, 10.0],
                    [90.0, 1.0, 3.0],
                    [90.0, 0.0, 0.0],
                    [80.0, 1.0, 4.0],
                    [50.0, 1.0, 3.0],
                    [40.0, 0.0, 0.0]])
    y_true = np.array([1.0, 1.0, 0.0, 1.0, 0.0, 0.0])
    X[:, 0] /= 100.0
    X[:, 2] /= 10.0
    
    # 2. Обучение
    w, b = train(X, y_true)
    
    # 3. Итоговая точность
    y_pred = predict(X, w, b)
    y_class = (y_pred >= 0.5).astype(int)
    accuracy = np.sum(y_class == y_true) / X.shape[0]
    print(f' Итоговая точность: {accuracy*100:.2f}%')
```

```python
example()
```
